```python
#accessing 

S= { 1 ,2 ,3 , 4 ,1 ,2 } 

if (10 in S):
    print("3 is present")
else:
    print("not present")

sum =0 

for element in S:
    sum+= element 

print(sum)
```

```python
#mutable

s= {5, 1 , 2 ,4, 3} 

s.add(10) 
s.add(0)

print(s)

s.pop()
s.pop()
print(s)

s.remove(5) 

print(s)
```

## Set Mathematical Methods

```python
a = { 1, 2, 3 } 
b = {1, 2 , 5,6, 4 } 

print( a.union(b) )

print(a.intersection(b)) 


print(a.isdisjoint(b)) 


print(a.issubset(b))
```
