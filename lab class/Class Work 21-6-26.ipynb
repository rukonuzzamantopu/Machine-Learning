```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Creating a sample dataset
data = {
    "Student_ID": [1, 2, 3, 4, 5, 6, 7, 8],
    "Study_Hours": [5, 2, 7, 1, 6, 3, 8, 4],
    "Attendance": [80, 55, 90, 40, 85, 60, 95, 70],
    "Previous_GPA": [3.20, 2.10, 3.70, 1.80, 3.40, 2.50, 3.90, 2.90],
    "Result": ["Pass", "Fail", "Pass", "Fail", "Pass", "Fail", "Pass", "Pass"]
}

# Converting dictionary into Pandas DataFrame
df = pd.DataFrame(data)

# Displaying the dataset
print("Dataset:")
print(df)
```

```python
print(df.shape)
```

```python
print(df.isnull().sum())
```

```python
# Displaying statistical summary
print("\nStatistical summary:")
print(df.describe())
```

```python
# Counting result values
print("\nResult value counts:")
print(df["Result"].value_counts())
```

```python
# Bar chart for result distribution
plt.figure(figsize=(6, 4))
sns.countplot(x="Result", data=df)
plt.title("Distribution of Student Results")
plt.xlabel("Result")
plt.ylabel("Number of Students")
plt.show()
```

```python
# Scatter plot between Study Hours and Previous GPA
plt.figure(figsize=(6, 4))
sns.scatterplot(x="Study_Hours", y="Previous_GPA", hue="Result", data=df)
plt.title("Study Hours vs Previous GPA")
plt.xlabel("Study Hours")
plt.ylabel("Previous GPA")
plt.show()
```
