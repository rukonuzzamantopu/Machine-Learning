```python
from google.colab import drive
drive.mount('/content/drive/')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

path = '/content/drive/MyDrive/CSE404DemoData/synthetic_dataset_with_missing_and_outliers.csv'
#Load the dataset using the Pandas library.
df = pd.read_csv(path)
#Display the first five rows of the dataset.
df.head()
#Display the shape of the dataset.
print(df.shape)
df.shape

#Display the column names and data types.
print(df.info())
```

```python
df['Gender'] = df['Gender'].replace({
    'M': 'Male',
    'm': 'Male',
    'male': 'Male',
    'MALE': 'Male',
     'F': 'Female',
    'f': 'Female',
    'female': 'Female',
    'FEMALE': 'Female'

})

print(df['Gender'].unique())
```

```python
print(df['City'].unique())
```

```python
df['City'] = df['City'].replace({
    'New York': 'New York',
    'New York City': 'New York',
    'new york' : 'New York',
    'LONDON' : 'London'})
print(df['City'].unique())
```

```python
print(df['Has_Degree'].unique())
```

```python
df['Has_Degree'] = df['Has_Degree'].replace({
    'YES': 'Yes',
    'no' : 'No'})
print(df['Has_Degree'].unique())
```

```python
print(df['Favorite_Color'].unique())
```

```python
df['favorite_color'] = df['Favorite_Color'].replace({
    'blue' :'Blue',
    'red' : 'Red',
    'RED' : 'Red',
    'green' : 'Green'
})
print(df['favorite_color'].unique())
```

```python
print(df.isnull().sum())
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
print(df.isnull().sum())
```

```python
df['Income'] = df['Income'].fillna(df['Income'].mean())
```
